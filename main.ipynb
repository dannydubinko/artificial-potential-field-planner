```python
import numpy as np
import matplotlib.pyplot as plt
from dataclasses import dataclass

```

```python
# Link to how I made 1 point move to another point
#https://gemini.google.com/app/02f1e47e3e21dfdf
```

```python
@dataclass
class Robot:
    x: float
    y: float

@dataclass
class Goal:
    x: float
    y: float

```

```python
# Create Time, Create steps
SIM_TIME = 10.0 # how long thing run for
T = 1 # time step, amount of time per node

num_nodes = int(round(SIM_TIME / T)) + 1
t = np.linspace(0, SIM_TIME, num_nodes) # generate time
N = np.size(t) # generate amount of "nodes"

# Set up Robot array and inital position
robot = np.zeros([2, N])
robot[:, 0] = 1.0


# Set Final Goal
goal = np.array([5.0, 5.0])
t
```

```python
for k in range(1, N):
    new_point = robot[:, 0] + (T*k)/SIM_TIME * (goal-robot[:, 0])
    robot[0][k] = new_point[0]
    robot[1][k] = new_point[1]
    print(k)
robot
```

```python
# 2. Create the plo
plt.plot(robot[0, :], robot[1, :], color='blue', marker='o', linestyle='-', label='Robot')
plt.plot(goal[0], goal[1], color='green', marker='o', linestyle='-', label='Goal')
plt.plot(robot[0, 0], robot[1, 0], color='red', marker='o', linestyle='-', label='Inital Position')

# 3. Add context and labels
plt.title("My First Matplotlib Plot", fontsize=14, fontweight='bold')
plt.xlabel("X Axis Label", fontsize=12)
plt.ylabel("Y Axis Label", fontsize=12)

# 4. Enhance readability
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# 5. Display the plot
plt.show()
```
